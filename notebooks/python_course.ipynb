## 1.1 Variables and dynamic typing

A **variable** is a name bound to an object. You create it by assignment (`=`); there are no declarations. Python is

- **dynamically typed**: the *object* has a type, the *name* does not, so a name can be re-bound to an object of another type,
- **strongly typed**: Python does not silently convert between unrelated types (`"1" + 1` is an error).

````python
price = 101.5  # float
ticker = "PKO"  # str
n_shares = 200  # int
is_listed = True  # bool

print(type(price), type(ticker), type(n_shares), type(is_listed))

# The same name can later point to an object of a different type (legal, but rarely a good idea)
x = 10
print(x, type(x))
x = "ten"
print(x, type(x))

# Multiple assignment and swapping without a temporary variable
bid, ask = 99.8, 100.2
bid, ask = ask, bid
print(f"bid={bid}, ask={ask}")
````

**Naming rules.** Names consist of letters, digits and `_`, cannot start with a digit and are case-sensitive. By convention (PEP 8): `snake_case` for variables and functions, `PascalCase` for classes, `UPPER_CASE` for constants. Keywords such as `for`, `class`, `lambda` cannot be used as names.

**Comments** start with `#`. Write comments that explain *why*, not *what*.

## 1.2 Numbers

Python has three built-in numeric types: `int` (arbitrary precision — it never overflows), `float` (64-bit IEEE-754 double) and `complex`.

| Operator | Meaning | Example |
|---|---|---|
| `+ - *` | addition, subtraction, multiplication | `3 * 4` → `12` |
| `/` | true division — **always returns a float** | `7 / 2` → `3.5` |
| `//` | floor division (rounds towards −∞) | `7 // 2` → `3`, `-7 // 2` → `-4` |
| `%` | remainder (sign follows the divisor) | `7 % 2` → `1`, `-7 % 2` → `1` |
| `**` | power | `2 ** 10` → `1024` |

````python
print("7 / 2   =", 7 / 2)
print("7 // 2  =", 7 // 2)
print("-7 // 2 =", -7 // 2, " (floor, not truncation!)")
print("7 % 2   =", 7 % 2)
print("2 ** 10 =", 2**10)
print("2 ** 0.5 =", 2**0.5)

# Integers have arbitrary precision
print("2 ** 200 =", 2**200)

# Underscores make large literals readable; scientific notation gives floats
notional = 25_000_000
print(notional, 1.5e6, 2e-4)

# Augmented assignment
cash = 1000
cash += 250  # same as cash = cash + 250
cash *= 1.05
print("cash =", cash)
````

### Pitfall: floating-point arithmetic

A `float` stores a binary fraction. Most decimal fractions (0.1, 0.2, 0.35, ...) have no exact binary representation, so tiny rounding errors are unavoidable. Consequences:

- never compare floats with `==`; use `math.isclose` (or `np.isclose` for arrays),
- `round()` uses **banker's rounding** (round-half-to-even) *and* operates on the binary value, so results may surprise you,
- for money, where cents must be exact, use `decimal.Decimal` (or integers representing cents).

````python
import math
from decimal import ROUND_HALF_UP, Decimal

print("0.1 + 0.2 =", 0.1 + 0.2)
print("0.1 + 0.2 == 0.3 ?", 0.1 + 0.2 == 0.3)
print("math.isclose(0.1 + 0.2, 0.3) ?", math.isclose(0.1 + 0.2, 0.3))

# Banker's rounding: exact halves go to the nearest EVEN number
print("round(0.5), round(1.5), round(2.5) =", round(0.5), round(1.5), round(2.5))
# 2.675 is stored as 2.67499999..., so it rounds DOWN
print("round(2.675, 2) =", round(2.675, 2))

# Decimal: exact decimal arithmetic - always construct it from a string!
amount = Decimal("2.675")
print("Decimal rounding:", amount.quantize(Decimal("0.01"), rounding=ROUND_HALF_UP))
print("Decimal('0.1') + Decimal('0.2') =", Decimal("0.1") + Decimal("0.2"))

# Special float values
print(float("inf") > 1e308, float("nan") == float("nan"), math.isnan(float("nan")))
````

## 1.3 Booleans, `None`, comparisons and logic

- `bool` has two values: `True` and `False` (they are in fact the integers 1 and 0, so `sum` of booleans counts the `True`s).
- `None` is the single "no value" object. Test for it with `is None` / `is not None`, never with `==`.
- Comparison operators: `== != < <= > >=`. They can be **chained**: `0 < x <= 1`.
- Logical operators are the words `and`, `or`, `not`. They **short-circuit**: the right-hand side is evaluated only if needed.
- `==` compares **values**, `is` compares **identity** (is it the very same object in memory?).

````python
rate = 0.035
print("0 < rate <= 0.05:", 0 < rate <= 0.05)
print("True + True =", True + True)

# Short-circuit: the division is never evaluated when n == 0
n = 0
print("safe check:", n != 0 and 10 / n > 1)

# == versus is
a = [1, 2, 3]
b = [1, 2, 3]
c = a
print("a == b:", a == b, "| a is b:", a is b, "| a is c:", a is c)

result = None
print("result is None:", result is None)
````

### Truthiness

Every object can be used in a condition. The following are **falsy**: `False`, `None`, zero (`0`, `0.0`), and **empty containers** (`""`, `[]`, `()`, `{}`, `set()`). Everything else is truthy. This allows the idiom `if my_list:` meaning "if the list is not empty".

`and` / `or` return **one of their operands**, not necessarily a `bool` — hence the idiom `value or default`.

````python
for obj in [0, 1, -1, 0.0, "", "0", [], [0], None, {}]:
    print(f"{obj!r:>6} -> {bool(obj)}")

name = ""
print("Hello,", name or "anonymous")
````

# 2. Strings

A `str` is an **immutable sequence of Unicode characters**. Literals use single or double quotes (no difference); triple quotes span several lines. The backslash introduces escape sequences (`\n` newline, `\t` tab); a raw string `r"..."` switches escapes off (useful for Windows paths and regular expressions).

## 2.1 Indexing and slicing

Indexing is **zero-based**; negative indices count from the end. A slice `s[start:stop:step]` includes `start` and **excludes** `stop`. The same rules apply to lists, tuples, NumPy arrays and pandas `.iloc`.

````python
isin = "PLPKO0000016"

print("first char:     ", isin[0])
print("last char:      ", isin[-1])
print("country code:   ", isin[:2])  # from the start up to (not including) index 2
print("without country:", isin[2:])
print("every 2nd char: ", isin[::2])
print("reversed:       ", isin[::-1])
print("length:         ", len(isin))

# Strings are immutable: item assignment raises TypeError
try:
    isin[0] = "X"
except TypeError as err:
    print("TypeError:", err)
````

## 2.2 Common string methods

Methods never modify the string (it is immutable); they return a **new** one.

````python
raw = "  Bank Pekao SA ; WSE ; banking  "

clean = raw.strip()  # remove leading/trailing whitespace
print(repr(clean))
print(clean.lower(), "|", clean.upper())
print(clean.replace(" ; ", ","))

parts = [p.strip() for p in raw.split(";")]  # split into a list
print(parts)
print(" | ".join(parts))  # join a list back into one string

print(clean.startswith("Bank"), clean.endswith("ing"), "WSE" in clean)
print(clean.find("WSE"), clean.count("a"))
print("123".isdigit(), "12.3".isdigit(), "abc".isalpha())
print("pko".center(11, "*"), "7".zfill(3))
````

## 2.3 f-strings and format specifications

An **f-string** (`f"..."`) evaluates the expressions in `{}` and inserts them into the text. After a colon you can give a *format specification*:

| Spec | Meaning | Example → result |
|---|---|---|
| `:.2f` | fixed-point, 2 decimals | `f"{3.14159:.2f}"` → `3.14` |
| `:,` / `:,.2f` | thousands separator | `f"{1234567.891:,.2f}"` → `1,234,567.89` |
| `:.1%` | percentage (multiplies by 100) | `f"{0.0345:.1%}"` → `3.5%` |
| `:.3e` | scientific notation | `f"{0.000123:.3e}"` → `1.230e-04` |
| `:>10` `:<10` `:^10` | align right / left / centre in 10 characters | |
| `:05d` | integer padded with zeros | `f"{42:05d}"` → `00042` |
| `{x=}` | debugging: prints `x=<value>` | |
| `!r` | use `repr()` instead of `str()` | |

You will see f-strings in every lab notebook, e.g. `print(f"Accuracy: {accuracy:.4f}")`.

````python
ticker, price, ret, volume = "PKN", 64.3, 0.01234, 1_234_567

print(f"{ticker}: price={price:.2f} PLN, return={ret:.2%}, volume={volume:,}")
print(f"{price=}, {ret * 100=:.3f}")

# Aligned table
rows = [("PKO", 51.2, 0.0123), ("PZU", 47.85, -0.0045), ("KGHM", 131.0, 0.0310)]
print(f"{'Ticker':<8}{'Price':>10}{'Return':>10}")
print("-" * 28)
for t, p, r in rows:
    print(f"{t:<8}{p:>10.2f}{r:>10.2%}")
````

## 2.4 `str` versus `bytes`

`str` is text (Unicode code points). `bytes` is raw binary data — what is actually stored in files and sent over a network. You go from one to the other by **encoding** (`str → bytes`) and **decoding** (`bytes → str`), almost always with UTF-8. This matters when a file with Polish characters is read with the wrong encoding.

````python
text = "Michał Woźniak, zł"
data = text.encode("utf-8")

print(data)
print(f"{len(text)} characters, {len(data)} bytes")  # non-ASCII characters take more than one byte
print(data.decode("utf-8"))
print(data.decode("latin-1"))  # wrong encoding -> garbage ("mojibake")
````

# 3. Collections

Python has four built-in container types. Choosing the right one makes code simpler and faster.

| Type | Literal | Ordered | Mutable | Duplicates | Typical use |
|---|---|---|---|---|---|
| `list` | `[1, 2, 3]` | yes | yes | yes | a sequence of items that changes |
| `tuple` | `(1, 2, 3)` | yes | **no** | yes | a fixed record, a function returning several values, a dict key |
| `dict` | `{"a": 1}` | yes (insertion order) | yes | keys unique | look-up by key |
| `set` | `{1, 2, 3}` | no | yes | **no** | membership tests, removing duplicates, set algebra |

## 3.1 Lists

````python
prices = [101.2, 99.8, 100.5, 102.1]

# Indexing and slicing work exactly as for strings
print(prices[0], prices[-1], prices[1:3])

# Lists are mutable
prices[0] = 101.0
prices.append(103.4)  # add one element at the end
prices.extend([104.0, 103.2])  # add several elements
prices.insert(1, 100.0)  # insert at position 1 (slow for long lists)
print(prices)

last = prices.pop()  # remove and return the last element
prices.remove(100.0)  # remove the first occurrence of a value
print("popped:", last, "|", prices)

print("len:", len(prices), "| max:", max(prices), "| index of 102.1:", prices.index(102.1))
print("is 99.8 in the list?", 99.8 in prices)

# Concatenation and repetition create NEW lists
print([1, 2] + [3], [0] * 5)
````

### Sorting: `sorted()` versus `.sort()`

- `sorted(iterable)` returns a **new** sorted list and works on any iterable,
- `list.sort()` sorts **in place** and returns `None` (a classic bug: `x = x.sort()` sets `x` to `None`).

Both accept `reverse=True` and `key=` — a function computing the value to sort by.

````python
tickers = ["pko", "KGHM", "Allegro", "cdr"]

print(sorted(tickers))  # upper-case letters sort before lower-case ones
print(sorted(tickers, key=str.lower))  # case-insensitive
print(sorted(tickers, key=len, reverse=True))  # by length, longest first
print("original is unchanged:", tickers)

tickers.sort(key=str.lower)
print("after in-place sort:  ", tickers)
````

## 3.2 Tuples and unpacking

A tuple is an immutable sequence. The parentheses are optional; it is the **comma** that makes a tuple (`(5,)` is a one-element tuple, `(5)` is just the number 5).

**Unpacking** assigns the elements of any iterable to several names at once. A starred name collects "the rest".

````python
trade = ("PKO", "BUY", 200, 51.2)
ticker, side, qty, px = trade  # unpacking
print(f"{side} {qty} x {ticker} @ {px}")

first, *middle, last = [1, 2, 3, 4, 5]  # starred unpacking
print(first, middle, last)

# Functions return several values as a tuple
quotient, remainder = divmod(17, 5)
print(quotient, remainder)

# The * operator also unpacks an iterable INTO a call or a literal
bounds = (0, 10, 2)
print(list(range(*bounds)))
print([*"abc", *[1, 2]])

# Tuples are immutable
try:
    trade[2] = 300
except TypeError as err:
    print("TypeError:", err)
````

## 3.3 Dictionaries

A `dict` maps **keys** to **values**. Keys must be *hashable* (immutable: strings, numbers, tuples of those); values can be anything. Look-up, insertion and deletion take constant time on average, no matter how large the dictionary is.

````python
position = {"ticker": "PZU", "qty": 500, "price": 47.85}

print(position["ticker"])  # look-up by key
position["qty"] += 100  # update a value
position["currency"] = "PLN"  # add a new key
print(position)

# A missing key raises KeyError; .get() returns a default instead
print(position.get("sector"), position.get("sector", "unknown"))
print("'qty' in position:", "qty" in position)  # `in` tests KEYS

# Iteration
for key, value in position.items():
    print(f"  {key:<10} -> {value}")
print(list(position.keys()), list(position.values()))

removed = position.pop("currency")
print("removed:", removed, "|", position)

# Merging: ** unpacks a dict into another dict literal; | is the merge operator (later keys win)
defaults = {"n_neighbors": 5, "weights": "uniform"}
overrides = {"weights": "distance"}
print({**defaults, **overrides})
print(defaults | overrides)
````

## 3.4 Sets

````python
portfolio_a = {"PKO", "PZU", "KGHM", "CDR"}
portfolio_b = {"PKO", "ALE", "CDR", "LPP"}

print("union:               ", portfolio_a | portfolio_b)
print("intersection:        ", portfolio_a & portfolio_b)
print("only in A:           ", portfolio_a - portfolio_b)
print("in exactly one:      ", portfolio_a ^ portfolio_b)
print("is {'PKO'} a subset? ", {"PKO"} <= portfolio_a)

# Removing duplicates (the order is NOT preserved by a set)
orders = ["PKO", "PZU", "PKO", "CDR", "PZU"]
print(set(orders), len(set(orders)))
# Removing duplicates while keeping the order of first appearance
print(list(dict.fromkeys(orders)))
````

### Complexity intuition

`x in some_list` scans the list element by element — time grows linearly with its length, $O(n)$. `x in some_set` (or `some_dict`) uses a hash table — constant time on average, $O(1)$. For a few elements it does not matter; for a million it is the difference between milliseconds and minutes.

````python
import time

n = 2_000_000
big_list = list(range(n))
big_set = set(big_list)

start = time.perf_counter()
found = (n - 1) in big_list  # worst case: the last element
t_list = time.perf_counter() - start

start = time.perf_counter()
found = (n - 1) in big_set
t_set = time.perf_counter() - start

print(f"list: {t_list * 1e3:.3f} ms | set: {t_set * 1e3:.5f} ms | found: {found}")
````

## 3.5 Pitfall: mutability, aliasing and copying

Assignment **never copies** an object; it only binds another name to the same object. With mutable objects (lists, dicts, sets, NumPy arrays, DataFrames) a change made through one name is visible through all of them. This is one of the most common sources of bugs.

- **Alias**: `b = a` — the same object.
- **Shallow copy**: `a.copy()`, `list(a)`, `a[:]` — a new outer container, but the inner objects are still shared.
- **Deep copy**: `copy.deepcopy(a)` — everything is copied recursively.

````python
import copy

weights = [0.5, 0.3, 0.2]
alias = weights  # NOT a copy
alias[0] = 0.9
print("weights changed through the alias:", weights)

shallow = weights.copy()
shallow[0] = 0.1
print("weights untouched by the copy:    ", weights)

# Nested structure: a shallow copy shares the inner lists
book = [["PKO", 100], ["PZU", 200]]
shallow = book.copy()
deep = copy.deepcopy(book)

book[0][1] = 999
print("shallow copy sees the change:", shallow[0])
print("deep copy does not:          ", deep[0])

# The same trap with list repetition: three references to ONE inner list
grid = [[0] * 2] * 3
grid[0][0] = 1
print("wrong:", grid)
grid = [[0] * 2 for _ in range(3)]  # three independent inner lists
grid[0][0] = 1
print("right:", grid)
````

## 4.2 `for` loops, `range`, `enumerate`, `zip`

````python
tickers = ["PKO", "PZU", "KGHM"]
weights = [0.5, 0.3, 0.2]

# range(start, stop, step): stop is excluded; it is lazy (does not build a list)
print(list(range(5)), list(range(2, 10, 3)), list(range(5, 0, -1)))

# Pythonic iteration: over elements, not over indices
for t in tickers:
    print(t, end=" ")
print()

# enumerate yields (index, element); start=1 for human-friendly numbering
for i, t in enumerate(tickers, start=1):
    print(f"{i}. {t}")

# zip pairs elements up; it stops at the SHORTEST input (strict=True raises an error instead)
for t, w in zip(tickers, weights, strict=True):
    print(f"{t}: {w:.0%}")

# zip is also the idiomatic way to build a dict from two lists
print(dict(zip(tickers, weights, strict=True)))
````

## 4.4 The `match` statement

`match` (Python ≥ 3.10) performs **structural pattern matching**: it compares a value against patterns and can unpack it at the same time. It is more than a `switch`: patterns can describe the *shape* of tuples, lists, dicts and objects.

````python
def describe(order):
    """Interpret an order given as a tuple."""
    match order:
        case ("BUY" | "SELL" as side, ticker, qty) if qty > 0:
            return f"market order: {side} {qty} {ticker}"
        case (side, ticker, qty, limit):
            return f"limit order: {side} {qty} {ticker} @ {limit}"
        case ("CANCEL", order_id):
            return f"cancel order #{order_id}"
        case _:  # wildcard: matches anything
            return "unknown order"


for o in [("BUY", "PKO", 100), ("SELL", "PZU", 50, 48.0), ("CANCEL", 17), ("HOLD",)]:
    print(describe(o))
````

## 4.5 Comprehensions

A comprehension builds a list, dict or set from an iterable in a single readable expression:

```python
[expression for item in iterable if condition]
```

It replaces the "create empty list → loop → append" pattern. Keep comprehensions short; if you need nested logic, write a normal loop.

A **generator expression** has the same syntax in round brackets. It produces items lazily, one at a time, without building a list in memory — ideal as an argument of `sum`, `max`, `any`, `all`, ...

````python
prices = [100.0, 101.5, 99.8, 102.3, 103.1]

# Loop version
simple_returns = []
for i in range(1, len(prices)):
    simple_returns.append(prices[i] / prices[i - 1] - 1)

# List comprehension version (zip of the list with itself shifted by one)
simple_returns = [p1 / p0 - 1 for p0, p1 in zip(prices[:-1], prices[1:], strict=True)]
print([f"{r:.2%}" for r in simple_returns])

# With a filter, and with a conditional expression
print([r for r in simple_returns if r > 0])
print(["up" if r > 0 else "down" for r in simple_returns])

# Dict and set comprehensions
tickers = ["PKO", "PZU", "KGHM", "PKN"]
print({t: len(t) for t in tickers})
print({t[0] for t in tickers})

# Generator expression: no intermediate list is created
print(sum(r**2 for r in simple_returns))
print(any(r < -0.01 for r in simple_returns), all(p > 0 for p in prices))

# Nested comprehension: read the for-clauses left to right, like nested loops
print([(i, j) for i in range(3) for j in range(i)])
````

# 5. Functions

A function packages a piece of logic under a name so that it can be reused and tested. Rule of thumb: if you copy-paste a block of code for the second time, turn it into a function.

## 5.1 Defining and calling functions

- `def name(parameters):` defines a function; `return` sends a value back (a function without `return` returns `None`),
- the string right below the `def` line is the **docstring** — it is what `help()` shows. The lab notebooks use the **NumPy docstring style** shown below,
- **type hints** (`x: float`, `-> float`) document the expected types. Python does **not** enforce them at run time; they help readers, IDEs and static checkers.

````python
def present_value(cash_flow: float, rate: float, periods: int = 1) -> float:
    """
    Discount a single future cash flow.

    Parameters:
    -----------
    cash_flow : float
        Amount received in the future
    rate : float
        Discount rate per period (0.05 means 5%)
    periods : int, default=1
        Number of periods until the cash flow is received

    Returns:
    --------
    float : Present value of the cash flow
    """
    return cash_flow / (1 + rate) ** periods


print(present_value(1000, 0.05))  # positional arguments, default periods=1
print(present_value(1000, 0.05, 10))  # all positional
print(present_value(1000, rate=0.05, periods=10))  # keyword arguments: clearer at the call site
print(present_value(periods=10, cash_flow=1000, rate=0.05))  # keywords can come in any order
````

### Returning several values

`return a, b` returns a tuple, which the caller usually unpacks.

````python
def mean_and_std(values):
    """Return the mean and the sample standard deviation of a list of numbers."""
    n = len(values)
    mean = sum(values) / n
    variance = sum((v - mean) ** 2 for v in values) / (n - 1)
    return mean, variance**0.5


mu, sigma = mean_and_std([0.01, -0.02, 0.015, 0.005])
print(f"mean={mu:.4f}, std={sigma:.4f}")
````

## 5.2 Pitfall: mutable default arguments

Default values are evaluated **once, when the function is defined**, not at every call. A mutable default (`[]`, `{}`) is therefore shared between calls. The standard fix is the `None` sentinel.

````python
def add_trade_buggy(trade, book=[]):  # the SAME list object is reused by every call
    book.append(trade)
    return book


print(add_trade_buggy("BUY PKO"))
print(add_trade_buggy("SELL PZU"))  # surprise: the first trade is still there


def add_trade(trade, book=None):
    if book is None:  # create a fresh list at CALL time
        book = []
    book.append(trade)
    return book


print(add_trade("BUY PKO"))
print(add_trade("SELL PZU"))
````

## 5.3 `*args` and `**kwargs`

- `*args` collects extra **positional** arguments into a tuple,
- `**kwargs` collects extra **keyword** arguments into a dict,
- parameters after a bare `*` are **keyword-only** (many library functions use this, e.g. `pd.read_csv(path, *, sep=...)`).

At the call site the same symbols do the opposite: `f(*my_list)` and `f(**my_dict)` unpack a collection into arguments.

````python
def portfolio_value(*positions, currency="PLN", **metadata):
    """Each position is a (quantity, price) tuple."""
    total = sum(qty * price for qty, price in positions)
    print(f"value: {total:,.2f} {currency} | positions: {len(positions)} | metadata: {metadata}")
    return total


portfolio_value((100, 51.2), (50, 47.85))
portfolio_value((10, 131.0), currency="EUR", owner="desk A", strategy="momentum")


def describe_strategy(*, window=20, weighting="equal"):  # keyword-only parameters
    return f"Strategy(window={window}, weighting={weighting})"


params = {"window": 60, "weighting": "risk-parity"}
print(describe_strategy(**params))  # dict -> keyword arguments

try:
    describe_strategy(60)
except TypeError as err:
    print("TypeError:", err)
````

## 5.4 Scope: the LEGB rule

When Python meets a name, it searches four scopes in order: **L**ocal (the current function) → **E**nclosing (outer functions) → **G**lobal (the module / notebook) → **B**uilt-in. Assignment inside a function always creates a **local** name, unless declared `global` or `nonlocal` (avoid both: functions should receive inputs as arguments and give results back with `return`).

In notebooks this matters: a function that silently reads a global variable will behave differently after you re-run some other cell.

````python
fee = 0.001  # global


def cost_global(amount):
    return amount * fee  # reads the GLOBAL fee - hidden dependency


def cost_local(amount):
    fee = 0.002  # a new LOCAL variable, the global one is untouched
    return amount * fee


print(cost_global(10_000), cost_local(10_000), fee)


# Better: make the dependency explicit
def cost(amount, fee_rate=0.001):
    return amount * fee_rate


print(cost(10_000, fee_rate=0.0015))
````

## 5.5 Functions are objects: lambdas and higher-order functions

Functions are ordinary objects: they can be stored in variables, put in lists and **passed to other functions**. A function that takes or returns a function is called *higher-order*. You have already used one: `sorted(..., key=...)`.

A `lambda` is a small anonymous function limited to a single expression: `lambda arguments: expression`. Use it for short throw-away `key=` functions; for anything longer, or anything you want to name, write a `def`.

````python
bonds = [
    {"name": "DS1033", "coupon": 0.0600, "maturity": 2033},
    {"name": "PS0729", "coupon": 0.0475, "maturity": 2029},
    {"name": "WS0447", "coupon": 0.0400, "maturity": 2047},
]

by_coupon = sorted(bonds, key=lambda b: b["coupon"], reverse=True)
print([b["name"] for b in by_coupon])
print(min(bonds, key=lambda b: b["maturity"])["name"])

# map / filter return lazy iterators; a comprehension is usually more readable
print(list(map(lambda b: b["name"].lower(), bonds)))
print(list(filter(lambda b: b["coupon"] > 0.045, bonds)))
print([b["name"].lower() for b in bonds if b["coupon"] > 0.045])


# Passing a function as an argument
def apply_to_returns(returns, func):
    return [func(r) for r in returns]


print(apply_to_returns([0.01, -0.02], abs))
print(apply_to_returns([0.01, -0.02], lambda r: round(r * 1e4)))  # in basis points
````

## 5.6 Closures and decorators

A function defined inside another function remembers the variables of the enclosing scope, even after the outer function has finished. Such a function is a **closure**.

A **decorator** is a function that takes a function and returns a modified version of it. The syntax `@decorator` above a `def` is shorthand for `func = decorator(func)`. Below we write a `@timer`, handy when comparing implementations in the labs. `functools.wraps` copies the name and the docstring of the original function onto the wrapper.

````python
def make_discounter(rate):
    """Return a function that discounts cash flows at a fixed rate (a closure)."""

    def discount(cash_flow, periods):
        return cash_flow / (1 + rate) ** periods  # `rate` is remembered

    return discount


discount_5 = make_discounter(0.05)
discount_8 = make_discounter(0.08)
print(f"{discount_5(1000, 10):.2f} vs {discount_8(1000, 10):.2f}")
````

````python
import functools
import time


def timer(func):
    """Decorator that prints the execution time of the decorated function."""

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        elapsed = time.perf_counter() - start
        print(f"[timer] {func.__name__} took {elapsed * 1e3:.2f} ms")
        return result

    return wrapper


@timer
def sum_of_squares(n):
    """Sum of squares 0^2 + 1^2 + ... + (n-1)^2."""
    return sum(i * i for i in range(n))


print(sum_of_squares(1_000_000))
print(sum_of_squares.__name__, "-", sum_of_squares.__doc__)
````

## 5.7 Recursion

A function may call itself. Every recursive function needs a **base case** that stops the recursion. Python limits the recursion depth (about 1000), so deep recursions should be rewritten as loops. Decision trees, which you will implement later in the course, are naturally recursive structures.

````python
def compound(capital, rate, years):
    """Capital after a number of years of annual compounding, computed recursively."""
    if years == 0:  # base case: stops the recursion
        return capital
    return compound(capital * (1 + rate), rate, years - 1)  # recursive case: a smaller problem


def binomial_paths(steps):
    """Number of distinct up/down paths in a binomial tree with a given number of steps."""
    if steps == 0:
        return 1
    return 2 * binomial_paths(steps - 1)


def flatten(nested):
    """Flatten an arbitrarily nested list."""
    flat = []
    for item in nested:
        if isinstance(item, list):
            flat.extend(flatten(item))  # recursive case
        else:
            flat.append(item)  # base case
    return flat


print(round(compound(1000, 0.05, 10), 2))
print(binomial_paths(10))
print(flatten([1, [2, [3, [4, 5]], 6], [[7]]]))
````

# 6. Errors and Debugging

## 6.1 Reading a traceback

When something goes wrong Python raises an **exception** and prints a **traceback**. Read it **from the bottom up**:

1. the **last line** gives the exception type and the message — *what* went wrong,
2. the lines above show the chain of calls, the most recent call last — *where* it went wrong. Look for the first frame that points into **your** code rather than into a library.

The most common exception types:

| Exception | Typical cause |
|---|---|
| `SyntaxError` / `IndentationError` | the code is not valid Python (missing colon, bracket, wrong indentation) |
| `NameError` | a name that does not exist (typo, or the defining cell was not run) |
| `TypeError` | an operation on an object of the wrong type (`"1" + 1`, wrong number of arguments) |
| `ValueError` | right type, inappropriate value (`int("abc")`, shapes that do not match) |
| `IndexError` / `KeyError` | index out of range / missing dictionary key or DataFrame column |
| `AttributeError` | the object has no such attribute or method (often: the object is `None`) |
| `ZeroDivisionError` | division by zero |
| `FileNotFoundError` | wrong path — check the working directory |
| `ModuleNotFoundError` | package not installed, or the wrong environment/kernel is selected |

````python
import traceback


def average_return(returns):
    return sum(returns) / len(returns)


def report(portfolio):
    return {name: average_return(r) for name, r in portfolio.items()}


# We catch the exception only to keep the notebook running; the printed traceback is what you would normally see.
try:
    report({"A": [0.01, 0.02], "B": []})
except ZeroDivisionError:
    print(traceback.format_exc())
    print("\n-> read from the bottom: ZeroDivisionError, raised in average_return, called from report")
````

## 6.2 Handling exceptions: `try` / `except` / `else` / `finally`

- `try`: the code that may fail,
- `except SomeError as err`: runs if that error (or a subclass) was raised. Catch **specific** exceptions; a bare `except:` hides bugs,
- `else`: runs only if **no** exception occurred,
- `finally`: runs **always** (clean-up).

````python
def parse_price(text):
    """Convert user input such as '101,50' to a float; return None if impossible."""
    try:
        value = float(text.replace(",", "."))
    except AttributeError:
        print(f"  {text!r}: not a string")
        return None
    except ValueError as err:
        print(f"  {text!r}: cannot parse ({err})")
        return None
    else:
        print(f"  {text!r}: ok")
        return value
    finally:
        print("  (finally always runs)")


for raw in ["101,50", "n/a", 17]:
    print(parse_price(raw))
````

## 6.3 Raising exceptions, custom exceptions and `assert`

- Use `raise` to signal that a function received input it cannot handle — **fail early and loudly** instead of returning nonsense.
- Define your own exception types by subclassing `Exception` when callers need to distinguish your errors.
- `assert condition, message` is a sanity check for the developer ("this can never happen"). Assertions can be switched off globally, so never use them to validate user input.

````python
class InsufficientFundsError(Exception):
    """Raised when an order exceeds the available cash."""


def buy(cash, price, qty):
    if price <= 0 or qty <= 0:
        raise ValueError(f"price and qty must be positive, got price={price}, qty={qty}")
    cost = price * qty
    if cost > cash:
        raise InsufficientFundsError(f"need {cost:,.2f}, have {cash:,.2f}")
    remaining = cash - cost
    assert remaining >= 0, "remaining cash can never be negative here"
    return remaining


for args in [(10_000, 50.0, 100), (10_000, 50.0, 300), (10_000, -5.0, 10)]:
    try:
        print("remaining cash:", buy(*args))
    except InsufficientFundsError as err:
        print("InsufficientFundsError:", err)
    except ValueError as err:
        print("ValueError:", err)
````

## 6.4 Debugging and timing in notebooks

**Debugging**

- The humble `print` (or `print(f"{x=}")`) is often enough. Check **types and shapes** first: `type(x)`, `len(x)`, `X.shape`, `df.dtypes`.
- `%debug` — run this magic in a new cell right after an exception to open the interactive debugger at the point of failure (`p var` prints a variable, `u`/`d` move up/down the call stack, `q` quits).
- `breakpoint()` — put it in your code to pause there and inspect variables (`n` next line, `c` continue, `q` quit).
- IDEs (VS Code, PyCharm) offer graphical debuggers for notebooks: click next to a line number to set a breakpoint.
- When a notebook behaves strangely, **restart the kernel and run all cells** — stale variables are the usual culprit.

**Timing** — IPython *magic commands* (they start with `%` and work only in IPython/Jupyter, not in plain Python files):

# 9. Object-Oriented Programming

Everything in Python is an **object**: a bundle of data (**attributes**) and functions that act on that data (**methods**). A **class** is the blueprint; an **instance** is a concrete object built from it. The data-science libraries are thoroughly object-oriented — a DataFrame, a figure, an array are all objects — and in the labs you will write your own classes as well.

## 9.1 Defining a class

- `__init__` is the initialiser: it runs when you create an instance, and sets up the instance attributes,
- `self` is the instance the method is called on. `obj.method(x)` is just a shorter way of writing `Class.method(obj, x)`,
- **instance attributes** (`self.x = ...`) belong to one object; **class attributes** (defined in the class body) are shared by all instances,
- `__repr__` defines how the object is displayed — invaluable when debugging.

````python
class Position:
    """A position in a single instrument."""

    currency = "PLN"  # class attribute: shared by all instances

    def __init__(self, ticker, qty, price):
        self.ticker = ticker  # instance attributes
        self.qty = qty
        self.price = price

    def market_value(self):
        return self.qty * self.price

    def update_price(self, new_price):
        self.price = new_price
        return self  # returning self allows method chaining

    def __repr__(self):
        return f"Position(ticker={self.ticker!r}, qty={self.qty}, price={self.price})"


pos = Position("PKO", 100, 51.2)
print(pos)  # uses __repr__
print(pos.ticker, pos.market_value(), pos.currency)
print(pos.update_price(52.0).market_value())  # chaining

# The method call is syntactic sugar:
print(Position.market_value(pos))

other = Position("PZU", 50, 47.85)
print(other.currency, "|", isinstance(other, Position), "|", type(other).__name__)
````

## 9.2 Properties

A **property** looks like an attribute from the outside but is computed (and optionally validated) by a method. Use it for derived quantities and for validation. A leading underscore (`_qty`) is the convention for "internal — do not touch from outside"; Python does not enforce privacy.

````python
class Loan:
    def __init__(self, principal, annual_rate):
        self.principal = principal
        self.annual_rate = annual_rate  # goes through the setter below

    @property
    def annual_rate(self):
        return self._annual_rate

    @annual_rate.setter
    def annual_rate(self, value):
        if not 0 <= value < 1:
            raise ValueError(f"annual_rate must be in [0, 1), got {value}")
        self._annual_rate = value

    @property
    def monthly_rate(self):  # read-only, derived
        return self._annual_rate / 12


loan = Loan(200_000, 0.072)
print(loan.annual_rate, round(loan.monthly_rate, 5))  # no parentheses: it is a property

try:
    loan.annual_rate = 7.2  # someone passed per cent instead of a fraction
except ValueError as err:
    print("ValueError:", err)
````

## 9.3 Inheritance and `super()`

A subclass **inherits** all attributes and methods of its parent, and may **override** them or add new ones. `super()` gives access to the parent's implementation. Inheritance expresses an *is-a* relation: a `CouponBond` *is an* `Instrument`.

Because both classes below provide `present_value`, code that works with a list of instruments does not need to know their concrete types — this is **polymorphism**.

````python
class Instrument:
    def __init__(self, name, face_value):
        self.name = name
        self.face_value = face_value

    def present_value(self, rate):
        raise NotImplementedError("subclasses must implement present_value")

    def __repr__(self):
        return f"{type(self).__name__}({self.name!r})"


class ZeroCouponBond(Instrument):
    def __init__(self, name, face_value, maturity):
        super().__init__(name, face_value)  # let the parent set name and face_value
        self.maturity = maturity

    def present_value(self, rate):
        return self.face_value / (1 + rate) ** self.maturity


class CouponBond(ZeroCouponBond):
    def __init__(self, name, face_value, maturity, coupon_rate):
        super().__init__(name, face_value, maturity)
        self.coupon_rate = coupon_rate

    def present_value(self, rate):
        coupon = self.face_value * self.coupon_rate
        pv_coupons = sum(coupon / (1 + rate) ** t for t in range(1, self.maturity + 1))
        return pv_coupons + super().present_value(rate)  # reuse the parent's discounting of the face value


book = [ZeroCouponBond("OK0728", 1000, 2), CouponBond("DS1033", 1000, 7, 0.06)]
for instrument in book:  # polymorphism: the same call, different behaviour
    print(f"{instrument!r:<28} PV @ 5% = {instrument.present_value(0.05):>9.2f}")

print(isinstance(book[1], Instrument), CouponBond.__mro__[:3])
````

## 9.4 Special ("dunder") methods

Methods with double underscores let your objects work with Python's built-in syntax: `__repr__` (display), `__eq__` (`==`), `__lt__` (`<`, enables sorting), `__len__` (`len()`), `__getitem__` (`obj[i]`), `__iter__` (looping), `__add__` (`+`), `__contains__` (`in`), `__call__` (`obj()`).

````python
class Portfolio:
    def __init__(self, positions=None):
        self.positions = list(positions) if positions else []

    def __len__(self):
        return len(self.positions)

    def __getitem__(self, index):
        return self.positions[index]

    def __iter__(self):
        return iter(self.positions)

    def __contains__(self, ticker):
        return any(p.ticker == ticker for p in self.positions)

    def __add__(self, other):
        return Portfolio(self.positions + other.positions)

    def __repr__(self):
        return f"Portfolio({[p.ticker for p in self.positions]})"

    def market_value(self):
        return sum(p.market_value() for p in self)


pf = Portfolio([Position("PKO", 100, 51.2), Position("PZU", 50, 47.85)])
pf2 = Portfolio([Position("KGHM", 10, 131.0)])
combined = pf + pf2

print(combined, "| len:", len(combined), "| first:", combined[0])
print("'PZU' in combined:", "PZU" in combined, "| value:", combined.market_value())
print(sorted((p.market_value() for p in combined), reverse=True))
````

## 9.5 Dataclasses

For classes that mainly hold data, `@dataclass` generates `__init__`, `__repr__` and `__eq__` from the annotated fields. `frozen=True` makes the instances immutable (and hashable); `order=True` adds comparison methods.

````python
from dataclasses import asdict, dataclass, field


@dataclass
class Trade:
    ticker: str
    qty: int
    price: float
    side: str = "BUY"
    tags: list[str] = field(default_factory=list)  # the correct way to have a mutable default

    @property
    def value(self) -> float:
        sign = 1 if self.side == "BUY" else -1
        return sign * self.qty * self.price


t1 = Trade("PKO", 100, 51.2)
t2 = Trade("PKO", 100, 51.2)
print(t1)
print("equal by value:", t1 == t2, "| same object:", t1 is t2)
print(asdict(t1), "| value:", t1.value)


@dataclass(frozen=True)
class Currency:
    code: str
    decimals: int = 2


pln = Currency("PLN")
try:
    pln.code = "EUR"
except AttributeError as err:  # FrozenInstanceError is a subclass of AttributeError
    print(type(err).__name__, "-", err)
````

# 10. Iterators and Generators

## 10.1 The iterator protocol

A `for` loop works with any **iterable**. Under the hood it calls `iter(obj)` to get an **iterator** and then calls `next(iterator)` repeatedly until `StopIteration` is raised. An iterator is *consumed* as you go: it can be traversed only once.

````python
prices = [100, 101, 103]

it = iter(prices)
print(next(it), next(it), next(it))
try:
    next(it)
except StopIteration:
    print("StopIteration: the iterator is exhausted")

# zip, map, filter, enumerate, open(...) and generator expressions are all one-shot iterators
pairs = zip("ab", [1, 2], strict=True)
print(list(pairs), list(pairs))  # the second list is empty!
````

## 10.2 Generators

A **generator function** contains `yield`. Calling it does not run the body; it returns a generator object. Each `next()` runs the body up to the next `yield`, hands over the value and **pauses**, keeping all local variables alive.

Generators are **lazy**: values are produced one at a time, only when requested. They can therefore represent huge or even infinite sequences in constant memory.

````python
import sys

import numpy as np


def countdown(n):
    print("  (generator started)")
    while n > 0:
        yield n
        n -= 1
    print("  (generator finished)")


gen = countdown(3)
print("created:", gen)
print(next(gen))
print(next(gen))
print(list(gen))  # consumes the rest


def random_walk(start, sigma, seed=0):
    """An INFINITE generator of prices following a Gaussian random walk."""
    rng = np.random.default_rng(seed)
    price = start
    while True:
        yield price
        price += rng.normal(0, sigma)


walk = random_walk(100.0, sigma=1.0)
print([round(next(walk), 2) for _ in range(5)])

# Laziness in numbers: a list of a million squares versus a generator expression
print(
    sys.getsizeof([i * i for i in range(1_000_000)]),
    "bytes vs",
    sys.getsizeof(i * i for i in range(1_000_000)),
    "bytes",
)
````

## 10.3 Example: rolling windows over a price series

A generator that yields consecutive windows is a natural building block for rolling statistics and for walk-forward backtests, where a rule is calibrated on a window of the past and evaluated on the following period.

````python
def rolling_windows(series, window):
    """Yield consecutive windows of a fixed length."""
    for end in range(window, len(series) + 1):
        yield series[end - window : end]


def walk_forward_splits(n_obs, train_size, test_size):
    """Yield (train_indices, test_indices) pairs that move forward in time."""
    start = 0
    while start + train_size + test_size <= n_obs:
        train = list(range(start, start + train_size))
        test = list(range(start + train_size, start + train_size + test_size))
        yield train, test
        start += test_size


closes = [100, 102, 101, 105, 107, 106, 110]
for w in rolling_windows(closes, window=3):
    print(w, "-> mean:", round(sum(w) / len(w), 2))

for train, test in walk_forward_splits(n_obs=10, train_size=4, test_size=2):
    print("train:", train, "| test:", test)
````

### Pitfall: views versus copies

A **slice of an array is a view**: it shares memory with the original, so writing to it changes the original (the opposite of lists, where a slice is a copy). Boolean and fancy indexing return **copies**. When in doubt, call `.copy()`.

````python
a = np.arange(6)
view = a[:3]  # a view on the first three elements
view[0] = 99
print("original changed through the view:", a)

safe = a[:3].copy()
safe[0] = -1
print("original untouched by the copy:  ", a)

picked = a[[0, 1]]  # fancy indexing -> copy
picked[0] = 12345
print("fancy indexing gave a copy:      ", a, "| shares memory:", np.shares_memory(a, picked))
````

## 11.4 Vectorisation

Python loops are slow because every iteration is interpreted. NumPy operations run in compiled C code over whole arrays. **Vectorisation** means replacing explicit loops with array expressions — typically 10–100 times faster and easier to read.

````python
import time

rng = np.random.default_rng(0)
p = 100 * np.exp(np.cumsum(rng.normal(0, 0.01, size=1_000_000)))  # a million prices


def log_returns_loop(prices):
    out = []
    for i in range(1, len(prices)):
        out.append(np.log(prices[i] / prices[i - 1]))
    return np.array(out)


def log_returns_vectorised(prices):
    return np.log(prices[1:] / prices[:-1])  # or simply np.diff(np.log(prices))


start = time.perf_counter()
r_loop = log_returns_loop(p[:200_000])
t_loop = time.perf_counter() - start

start = time.perf_counter()
r_vec = log_returns_vectorised(p[:200_000])
t_vec = time.perf_counter() - start

print(f"loop: {t_loop * 1e3:.1f} ms | vectorised: {t_vec * 1e3:.2f} ms | speed-up: {t_loop / t_vec:.0f}x")
print("same result:", np.allclose(r_loop, r_vec))
````

## 11.5 Aggregations and the `axis` argument

Aggregations (`sum`, `mean`, `std`, `min`, `max`, `argmax`, `cumsum`, ...) reduce an array. With `axis` you choose the dimension that is **collapsed**:

- `axis=0` → collapse the rows → one result **per column** (per feature),
- `axis=1` → collapse the columns → one result **per row** (per sample).

Note that `np.std` uses `ddof=0` (population formula, divide by $n$) by default, whereas pandas uses `ddof=1` (sample formula, divide by $n-1$).

````python
R = rng.normal(0.0005, 0.01, size=(250, 3))  # 250 daily returns of 3 assets

print("overall mean:       ", R.mean().round(5))
print("mean per asset:     ", R.mean(axis=0).round(5), "-> shape", R.mean(axis=0).shape)
print("mean per day:       shape", R.mean(axis=1).shape)
print("std (ddof=0 / 1):   ", R[:, 0].std().round(6), R[:, 0].std(ddof=1).round(6))
print("best day per asset: ", R.argmax(axis=0))
print("final cumulative return per asset:", (np.cumprod(1 + R, axis=0)[-1] - 1).round(4))
print("5% quantile (1-day historical VaR) per asset:", np.quantile(R, 0.05, axis=0).round(4))
````

## 11.6 Broadcasting

Broadcasting lets NumPy combine arrays of **different shapes** without copying data. Shapes are compared **from the right**; two dimensions are compatible when they are equal or one of them is 1 (a missing dimension counts as 1). The size-1 dimension is virtually stretched.

```
X        (250, 3)
X.mean(axis=0)  (3,)   -> treated as (1, 3) -> stretched to (250, 3)   OK
X.mean(axis=1)  (250,) -> treated as (1, 250) vs (250, 3)              ERROR
X.mean(axis=1, keepdims=True)  (250, 1) -> stretched to (250, 3)       OK
```

The most important application in this course: **standardisation** (z-scores) of every feature — exactly what `StandardScaler` does.

````python
X = rng.normal(loc=[50, 5000, 0.3], scale=[10, 1500, 0.1], size=(250, 3))  # 3 features on different scales

X_std = (X - X.mean(axis=0)) / X.std(axis=0)  # (250, 3) op (3,) -> broadcast over the rows
print("means after: ", X_std.mean(axis=0).round(10))
print("stds after:  ", X_std.std(axis=0).round(10))

# Demeaning each ROW needs keepdims=True (or [:, None]) to get shape (250, 1)
row_demeaned = X - X.mean(axis=1, keepdims=True)
print("row means after:", row_demeaned.mean(axis=1)[:3].round(10))

try:
    X - X.mean(axis=1)
except ValueError as err:
    print("ValueError:", err)

# Outer operations via broadcasting: (3, 1) with (1, 4) -> (3, 4)
strikes = np.array([90, 100, 110])[:, None]
spots = np.array([80, 95, 105, 120])[None, :]
print("call payoffs (rows: strikes, columns: spot prices):\n", np.maximum(spots - strikes, 0))
````

## 11.7 Linear algebra: solving a linear system

`@` is matrix multiplication (`*` is element-wise!), `.T` the transpose. A system $A x = b$ should be solved with `np.linalg.solve` rather than by computing $A^{-1}$ explicitly — faster and numerically more stable. When the system has more equations than unknowns, `np.linalg.lstsq` returns the least-squares solution.

Example — **cash-flow matching**: a pension fund must pay 100, 100 and 1,100 in years 1, 2 and 3. Three bonds are available (face value 100): a 1-year zero-coupon bond, a 2-year 5% coupon bond and a 3-year 4% coupon bond. How many units of each bond replicate the liabilities exactly?

````python
# Rows = years, columns = bonds; entry = cash flow of one unit of the bond in that year
cash_flows = np.array(
    [
        [100.0, 5.0, 4.0],  # year 1
        [0.0, 105.0, 4.0],  # year 2
        [0.0, 0.0, 104.0],  # year 3
    ]
)
liabilities = np.array([100.0, 100.0, 1100.0])

units = np.linalg.solve(cash_flows, liabilities)
print("units of each bond:", units.round(4))
print("liabilities replicated:", np.allclose(cash_flows @ units, liabilities))

# A rectangular system: fit a line through noisy points by least squares (no inverse needed)
x = np.linspace(0, 10, 50)
y = 3.0 + 0.5 * x + rng.normal(scale=0.4, size=50)
A = np.column_stack([np.ones_like(x), x])
(intercept, slope), *_ = np.linalg.lstsq(A, y, rcond=None)
print(f"least-squares line: intercept {intercept:.3f}, slope {slope:.3f}  (true values 3.0 and 0.5)")
````

## 11.8 Random numbers and simulation: geometric Brownian motion

Create a generator with `rng = np.random.default_rng(seed)` and draw from it (`rng.normal`, `rng.uniform`, `rng.integers`, `rng.choice`, ...). A fixed seed makes the results reproducible. (The older `np.random.seed` / `np.random.randn` interface still works but is no longer recommended for new code.)

Under geometric Brownian motion the price evolves as

$$S_{t+\Delta t} = S_t \exp\!\left[\left(\mu - \tfrac{1}{2}\sigma^2\right)\Delta t + \sigma\sqrt{\Delta t}\; Z_t\right], \qquad Z_t \sim N(0, 1).$$

Thanks to vectorisation we simulate thousands of paths without a single loop.

````python
def simulate_gbm(s0, mu, sigma, n_days, n_paths, seed=None):
    """
    Simulate daily price paths of a geometric Brownian motion.

    Returns:
    --------
    array of shape (n_days + 1, n_paths) : Simulated prices; row 0 contains s0
    """
    rng = np.random.default_rng(seed)
    dt = 1 / 252
    z = rng.standard_normal(size=(n_days, n_paths))
    log_increments = (mu - 0.5 * sigma**2) * dt + sigma * np.sqrt(dt) * z
    log_paths = np.vstack([np.zeros(n_paths), np.cumsum(log_increments, axis=0)])
    return s0 * np.exp(log_paths)


paths = simulate_gbm(s0=100, mu=0.08, sigma=0.20, n_days=252, n_paths=10_000, seed=42)
terminal = paths[-1]
print("shape:", paths.shape)
print(f"mean terminal price: {terminal.mean():.2f} (theory: {100 * np.exp(0.08):.2f})")
print(f"P(loss after 1 year): {(terminal < 100).mean():.1%}")
print(f"5% quantile of the 1-year return: {np.quantile(terminal / 100 - 1, 0.05):.1%}")
````

## 11.9 Returns, volatility, Sharpe ratio and portfolio variance

With daily log returns $r_t$: annualised volatility $= \operatorname{std}(r)\sqrt{252}$, annualised Sharpe ratio $= \dfrac{\bar{r} - r_f/252}{\operatorname{std}(r)}\sqrt{252}$. For portfolio weights $w$ and covariance matrix $\Sigma$ the portfolio variance is $w^\top \Sigma\, w$.

````python
# Three correlated assets: draw correlated normal returns with a given covariance matrix
vols = np.array([0.15, 0.25, 0.35]) / np.sqrt(252)  # daily volatilities
corr = np.array([[1.0, 0.6, 0.3], [0.6, 1.0, 0.5], [0.3, 0.5, 1.0]])
cov_true = np.outer(vols, vols) * corr
daily = rng.multivariate_normal(mean=[0.0003, 0.0004, 0.0006], cov=cov_true, size=1260)  # 5 years

ann_vol = daily.std(axis=0, ddof=1) * np.sqrt(252)
sharpe = (daily.mean(axis=0) - 0.03 / 252) / daily.std(axis=0, ddof=1) * np.sqrt(252)
print("annualised volatility:", ann_vol.round(3))
print("annualised Sharpe:    ", sharpe.round(2))

Sigma = np.cov(daily, rowvar=False) * 252  # rowvar=False: the COLUMNS are the variables
w = np.array([0.5, 0.3, 0.2])
port_vol = np.sqrt(w @ Sigma @ w)
print(f"portfolio volatility: {port_vol:.3f} vs weighted average of volatilities: {w @ ann_vol:.3f} (diversification!)")
print("correlation matrix:\n", np.corrcoef(daily, rowvar=False).round(2))
````

| `df[["a", "b"]]` | several columns (note the list) | `DataFrame` |
| `df[mask]` | rows where the boolean mask is `True` | `DataFrame` |
| `df.loc[rows, cols]` | by **label** (slices **include** the end) | |
| `df.iloc[rows, cols]` | by **integer position** (slices exclude the end, as in NumPy) | |
| `df.query("expr")` | rows satisfying an expression written as a string | `DataFrame` |

Conditions are combined with `&`, `|`, `~` and **parentheses**, exactly as in NumPy (`and` / `or` do not work on arrays). Useful helpers: `.isin([...])`, `.between(a, b)`, `.str.contains(...)`, `.isna()`.

````python
print(loans["amount"].head(3), "\n")
print(loans[["age", "income"]].head(3), "\n")

# Boolean filtering
big_young = loans[(loans["amount"] >= 50_000) & (loans["age"] < 30)]
print("big loans to young clients:", len(big_young))
print(loans[loans["purpose"].isin(["car", "education"]) & loans["income"].between(5_000, 6_000)].shape)

# The same with query(); @name refers to a Python variable
threshold = 50_000
print(len(loans.query("amount >= @threshold and age < 30")))

# loc: labels | iloc: positions
by_id = loans.set_index("loan_id")
print(by_id.loc[3:5, ["age", "amount"]])  # label slice 3:5 INCLUDES 5
print(by_id.iloc[2:5, [0, 2]])  # position slice 2:5 EXCLUDES 5

# loc with a mask and a column list: the most common pattern
print(loans.loc[loans["default"] == 1, ["age", "income", "amount"]].mean().round(1))
````

## 12.4 Creating and modifying columns

Operations on columns are **vectorised** like in NumPy and are **aligned on the index**. Create columns with `df["new"] = ...` or, in method chains, with `.assign(...)`. To set values on a subset of rows use **one** `.loc[mask, column] = value` call.

````python
loans["annual_income"] = loans["income"] * 12
loans["lti"] = loans["amount"] / loans["annual_income"]  # loan-to-income ratio
loans["log_amount"] = np.log(loans["amount"])

# Conditional columns
loans["is_large"] = loans["amount"] >= 50_000
loans["age_group"] = pd.cut(loans["age"], bins=[20, 30, 45, 60, 70], labels=["21-30", "31-45", "46-60", "61-70"])
loans["risk_flag"] = np.where((loans["lti"] > 0.8) & (loans["age"] < 30), "high", "normal")

# Set values for a subset of rows with a single .loc call
loans.loc[loans["purpose"] == "education", "risk_flag"] = "student"

print(loans[["amount", "annual_income", "lti", "is_large", "age_group", "risk_flag"]].head())
print(loans["risk_flag"].value_counts())
````

### Copy-on-write (pandas 3)

In pandas 3 every object derived from a DataFrame (a filtered subset, a selected column, ...) **behaves like a copy**: modifying it never changes the original. Two practical consequences:

1. Modifying a subset is safe and silent — and does **not** propagate back to the parent.
2. **Chained assignment never works**: `df[df["a"] > 0]["b"] = 1` or `df["b"][0] = 1` modify a temporary object and leave `df` unchanged (pandas emits a `ChainedAssignmentError` warning). Always use `df.loc[mask, "b"] = 1`.

For the same reason, prefer `df = df.method(...)` over the old `inplace=True` idiom.

````python
subset = loans[loans["amount"] == 5_000]
subset["amount"] = -1  # modifies only `subset`
print("original untouched:", (loans["amount"] == -1).sum(), "rows changed in `loans`")

# The right way to modify the original: a single .loc call on the original
demo = loans[["loan_id", "amount"]].copy()
demo.loc[demo["amount"] == 5_000, "amount"] = 5_500
print(demo["amount"].value_counts().sort_index().head(3))
````

## 12.5 Missing data

Missing values are shown as `NaN` (`NaT` for dates). Detect them with `isna()` / `notna()`; remove with `dropna()`; fill with `fillna()`, `ffill()` / `bfill()` (time series) or `interpolate()`. Most aggregations **skip** missing values by default.

## 12.8 Group-by: split – apply – combine

`df.groupby(keys)` **splits** the rows into groups, **applies** a function to each group and **combines** the results.

- `agg` reduces each group to one row. *Named aggregation* — `new_name=("column", "function")` — gives clean column names.
- `transform` returns a result of the **same length** as the input (each row receives its group's statistic) — ideal for feature engineering.

````python
summary = (
    loans.groupby("purpose")
    .agg(
        n_loans=("loan_id", "count"),
        avg_amount=("amount", "mean"),
        median_income=("income", "median"),
        default_rate=("default", "mean"),
    )
    .sort_values("default_rate", ascending=False)
    .round(3)
)
print(summary, "\n")

# Two keys -> a MultiIndex; unstack() pivots the inner level into columns
print(loans.groupby(["region", "age_group"], observed=True)["default"].mean().unstack().round(3), "\n")

# transform: a group statistic for every row (here: amount relative to the mean of the same purpose)
loans["amount_vs_purpose_mean"] = loans["amount"] / loans.groupby("purpose")["amount"].transform("mean")
print(loans[["purpose", "amount", "amount_vs_purpose_mean"]].head(4).round(2))
````

## 12.9 Combining tables: `merge` and `concat`

- `pd.merge(left, right, on=..., how=...)` is a SQL-style join. `how`: `"inner"` (keys present in both), `"left"` (all rows of the left table), `"right"`, `"outer"`. Always check the number of rows afterwards; `validate="many_to_one"` makes pandas check the key uniqueness for you.
- `pd.concat([df1, df2])` stacks tables vertically (or side by side with `axis=1`).

````python
regions = pd.DataFrame(
    {
        "region": ["Mazowieckie", "Malopolskie", "Slaskie", "Lubelskie"],
        "unemployment": [0.041, 0.046, 0.037, 0.072],
    }
)

merged = loans[["loan_id", "region", "default"]].merge(regions, on="region", how="left", validate="many_to_one")
print("rows before/after:", len(loans), len(merged))
print("loans without regional data (Pomorskie is missing on the right):", merged["unemployment"].isna().sum())

inner = loans[["loan_id", "region"]].merge(regions, on="region", how="inner")
print("inner join keeps only the matching rows:", len(inner))

# concat: stack two batches of loans
batch_1, batch_2 = loans.iloc[:3][["loan_id", "amount"]], loans.iloc[3:5][["loan_id", "amount"]]
print(pd.concat([batch_1, batch_2], ignore_index=True))
````

## 12.11 `apply` versus vectorised operations

`apply` calls a Python function for every element or row — flexible but **slow** (it is a hidden loop). Use it only when no vectorised alternative exists.

````python
import time


def monthly_instalment(row):
    """Annuity instalment for a 5-year loan at 9% per year."""
    r, n = 0.09 / 12, 60
    return row["amount"] * r / (1 - (1 + r) ** -n)


start = time.perf_counter()
slow = loans.apply(monthly_instalment, axis=1)  # row by row
t_apply = time.perf_counter() - start

start = time.perf_counter()
r, n = 0.09 / 12, 60
fast = loans["amount"] * r / (1 - (1 + r) ** -n)  # whole column at once
t_vec = time.perf_counter() - start

print(f"apply: {t_apply * 1e3:.1f} ms | vectorised: {t_vec * 1e3:.2f} ms | same result: {np.allclose(slow, fast)}")

# Element-wise mapping of a column with a dict is a good use of .map
print(loans["purpose"].map({"car": "secured", "renovation": "secured"}).fillna("unsecured").value_counts().to_dict())
````

## 12.12 Method chaining

Most DataFrame methods return a new DataFrame, so operations can be **chained** into a readable pipeline (wrap the chain in parentheses to break lines). Each step reads like a sentence, and no intermediate variables pollute the namespace.

````python
report = (
    loans.query("amount >= 10_000")
    .assign(
        lti=lambda d: d["amount"] / (12 * d["income"]),
        lti_bucket=lambda d: pd.qcut(d["lti"], q=4, labels=["Q1", "Q2", "Q3", "Q4"]),
    )
    .dropna(subset=["lti_bucket"])
    .groupby("lti_bucket", observed=True)
    .agg(n=("loan_id", "size"), avg_lti=("lti", "mean"), default_rate=("default", "mean"))
    .round(3)
)
print(report)
````

## 12.13 Time series

With a `DatetimeIndex` pandas becomes a time-series tool: date-based slicing, `shift`, `pct_change`, `rolling`, `expanding`, `resample`. We simulate two years of daily prices for three assets.

**Look-ahead bias.** When building features for prediction, the row for day $t$ may only contain information available *before* the prediction is made. `shift(1)` moves data one step into the future (yesterday's value appears in today's row) — the basic tool for creating lagged features safely.

````python
dates = pd.bdate_range("2024-01-01", periods=504)  # business days
rng_ts = np.random.default_rng(11)
daily_vol = np.array([0.012, 0.018, 0.025])
log_r = rng_ts.normal(loc=0.0004, scale=daily_vol, size=(len(dates), 3))
prices = pd.DataFrame(100 * np.exp(np.cumsum(log_r, axis=0)), index=dates, columns=["BANK", "ENERGY", "TECH"])

print(prices.head(3).round(2), "\n")
print(prices.loc["2024-03"].shape, "| partial-string slicing:", prices.loc["2024-03-01":"2024-03-08"].shape)

returns = prices.pct_change().dropna()  # simple returns
log_returns = np.log(prices / prices.shift(1)).dropna()  # log returns
print(returns.head(3).round(4), "\n")

# Rolling statistics: 21-day annualised volatility; expanding: all the history up to each date
rolling_vol = returns.rolling(window=21).std() * np.sqrt(252)
print(rolling_vol.dropna().tail(3).round(3), "\n")

# Resampling: change the frequency ("ME" = month end, "W" = week)
monthly_close = prices.resample("ME").last()
monthly_returns = monthly_close.pct_change().dropna()
print(monthly_returns.head(3).round(4), "\n")

# A lagged feature without look-ahead bias: yesterday's return explains today's
features = pd.DataFrame({"ret_today": returns["TECH"], "ret_lag1": returns["TECH"].shift(1)}).dropna()
print(features.head(3).round(4))
````

### Drawdowns

The drawdown at time $t$ is the fall from the running maximum: $DD_t = P_t / \max_{s \le t} P_s - 1$. With `cummax()` it is a one-liner, for all assets at once.

````python
running_max = prices.cummax()
drawdown = prices / running_max - 1

stats = pd.DataFrame(
    {
        "ann_return": returns.mean() * 252,
        "ann_vol": returns.std() * np.sqrt(252),
        "max_drawdown": drawdown.min(),
        "worst_day": drawdown.idxmin(),
    }
)
stats["sharpe"] = stats["ann_return"] / stats["ann_vol"]
print(stats.to_string(float_format="{:.3f}".format))
````

# 14. Writing Good Code

## 14.1 Style: PEP 8 in one minute

Code is read far more often than it is written. [PEP 8](https://peps.python.org/pep-0008/) is the community style guide:

- 4 spaces per indentation level; never mix tabs and spaces,
- `snake_case` for variables and functions, `PascalCase` for classes, `UPPER_CASE` for constants,
- spaces around binary operators and after commas: `total = price * qty`, `f(a, b)`; no spaces around `=` in keyword arguments: `f(x, scale=True)`,
- descriptive names: `default_rate`, not `dr` or `x2` (short names are fine for indices and well-known maths symbols such as `X`, `y`),
- one statement per line; imports at the top, one module per line.

You do not have to memorise the details: a **formatter** and a **linter** do the work. This repository uses [ruff](https://docs.astral.sh/ruff/) (configuration in `pyproject.toml`):

## 14.2 Structure

- **Small functions that do one thing.** If you cannot describe a function in one sentence without "and", split it.
- **No hidden state.** Functions take inputs as arguments and `return` outputs; they do not read or modify global variables.
- **DRY — don't repeat yourself.** Copy-pasted code multiplies bugs; write a function or a loop.
- **No magic numbers.** `TRADING_DAYS = 252` at the top beats a bare `252` scattered through the code.
- **Docstrings** for every non-trivial function: what it does, parameters, returns.
- **Fail early** with clear error messages (`raise ValueError(...)`) instead of propagating nonsense.

## 14.3 Reproducibility

- **Seeds.** Anything random — simulations, samples, shuffles — must be seeded: `np.random.default_rng(42)`, `random_state=42` wherever a library offers it. The Kaggle League requires notebooks that reproduce the submitted file exactly.
- **Environment.** `pyproject.toml` + `uv.lock` pin the package versions; `uv sync` recreates the environment anywhere.
- **Relative paths.** `Path("data") / "train.csv"`, never `C:\Users\me\Desktop\...`.
- **Never edit raw data by hand.** Every transformation lives in code.

````python
# The same computation twice: hard to read and fragile ...
import numpy as np

d = np.random.default_rng(0).normal(0.0005, 0.01, 750)
print(d.mean() / d.std(ddof=1) * 252**0.5, (np.cumprod(1 + d) / np.maximum.accumulate(np.cumprod(1 + d)) - 1).min())

# ... versus named constants, small documented functions, explicit inputs and outputs
TRADING_DAYS = 252


def sharpe_ratio(daily_returns: np.ndarray, risk_free_rate: float = 0.0) -> float:
    """Annualised Sharpe ratio of a series of daily simple returns."""
    excess = daily_returns - risk_free_rate / TRADING_DAYS
    return excess.mean() / excess.std(ddof=1) * np.sqrt(TRADING_DAYS)


def max_drawdown(daily_returns: np.ndarray) -> float:
    """Maximum drawdown (a negative number) of the wealth path implied by daily simple returns."""
    wealth = np.cumprod(1 + daily_returns)
    running_max = np.maximum.accumulate(wealth)
    return (wealth / running_max - 1).min()


strategy_returns = np.random.default_rng(0).normal(0.0005, 0.01, 750)
print(f"Sharpe: {sharpe_ratio(strategy_returns):.2f} | max drawdown: {max_drawdown(strategy_returns):.1%}")
````

TRADING_DAYS = 252


def performance_table(prices):
    """
    Summarise the performance of each price series.

    Parameters:
    -----------
    prices : DataFrame
        Prices indexed by date, one column per asset

    Returns:
    --------
    DataFrame : One row per asset with annualised return, volatility, Sharpe ratio,
        maximum drawdown and the date on which it occurred
    """
    r = prices.pct_change().dropna()
    dd = prices / prices.cummax() - 1
    table = pd.DataFrame(
        {
            "ann_return": r.mean() * TRADING_DAYS,
            "ann_vol": r.std() * np.sqrt(TRADING_DAYS),
            "max_drawdown": dd.min(),
            "max_dd_date": dd.idxmin(),
        }
    )
    table.insert(2, "sharpe", table["ann_return"] / table["ann_vol"])
    return table

# 4. Portfolio class
class Portfolio:
    """A portfolio with fixed weights, rebalanced daily."""

    def __init__(self, weights, name="portfolio"):
        if abs(sum(weights.values()) - 1) > 1e-8:
            raise ValueError(f"weights must sum to 1, got {sum(weights.values()):.6f}")
        self.weights = pd.Series(weights, dtype=float)
        self.name = name

    def returns(self, asset_returns):
        return (asset_returns[self.weights.index] * self.weights).sum(axis=1).rename(self.name)

    def volatility(self, asset_returns):
        sigma = asset_returns[self.weights.index].cov().to_numpy() * TRADING_DAYS
        w = self.weights.to_numpy()
        return float(np.sqrt(w @ sigma @ w))

    def __repr__(self):
        inside = ", ".join(f"{asset}={weight:.1%}" for asset, weight in self.weights.items())
        return f"Portfolio({self.name}: {inside})"


# 5. Equal-weight and minimum-variance portfolios: estimate on years 1-2, evaluate on year 3
in_sample, out_of_sample = asset_returns.iloc[:504], asset_returns.iloc[504:]

assets = list(market.columns)
equal_weight = Portfolio(dict.fromkeys(assets, 1 / len(assets)), name="equal_weight")

sigma_in = in_sample.cov().to_numpy()
raw = np.linalg.solve(sigma_in, np.ones(len(assets)))
min_variance = Portfolio(dict(zip(assets, raw / raw.sum(), strict=True)), name="min_variance")
print(equal_weight)
print(min_variance)

wealth = pd.DataFrame({p.name: 100 * (1 + p.returns(out_of_sample)).cumprod() for p in (equal_weight, min_variance)})
print(performance_table(wealth).to_string(float_format="{:.3f}".format))
for p in (equal_weight, min_variance):
    print(f"{p.name}: out-of-sample volatility via w'Sigma w = {p.volatility(out_of_sample):.3f}")

**Reflection.** The minimum-variance portfolio is dominated by the low-volatility `BOND` asset (about 90%) and has a far lower out-of-sample volatility than the equally weighted portfolio (roughly 6% versus 17%). In this particular simulated year it also earned more, but that is the luck of a single path, not a property of the method: by construction `BOND` has the lowest expected return, so ex ante low risk is paid for with a low expected return. Estimating $\Sigma$ on the first two years only mimics real life: on the day you choose the weights you do not know future returns; using the full sample would be *look-ahead bias* — the portfolio-construction analogue of fitting a scaler on the test set. Here all weights happen to be positive; a negative weight would be a short position (the closed-form solution does not rule it out). In practice you would often add a no-short-selling constraint, which requires a numerical optimiser (`scipy.optimize.minimize`) instead of the closed-form solution.
